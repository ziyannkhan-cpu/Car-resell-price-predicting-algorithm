```python
x=data.drop(columns=['selling_price','Unnamed: 0',],axis=1)
```

```python
data
```

```python
x=pd.get_dummies(x,columns=['fuel','seller_type','transmission'],drop_first=True)
x=x.astype(int)
```

```python
y=data['selling_price']
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
c=['km_driven','owner','engine','max_power','seats','age']
```

```python
x[c]=scaler.fit_transform(x[c])
```

```python
x
```

```python
from scipy.stats import chi2_contingency
alpha=0.05
cat_feat=['fuel_Diesel','fuel_LPG','fuel_Petrol','seller_type_Individual','seller_type_Trustmark Dealer','transmission_Manual','owner','seats','age']
data2['selling_price']=pd.qcut(data2['selling_price'],q=3,labels=['low','medium','high'])
```

```python
chi2_results={}
for cols in cat_feat:
    contingency=pd.crosstab(x[cols],data2['selling_price'])
    chi_stat,p_val,_,_=chi2_contingency(contingency)
    decision='reject null(keep feature)'if p_val<alpha else 'Accept null(drop feature)'
    chi2_results[cols]={'chi2_statistic':chi_stat,'p_value':p_val,'Decison':decision}
    chi2_results
```

```python
chi2_df=pd.DataFrame(chi2_results)
chi2_df
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.33, random_state=42)
```

```python
data.isnull().sum()
```

```python
from sklearn.ensemble import RandomForestRegressor
rf_model = RandomForestRegressor(n_estimators=100).fit(X_train, y_train)
y_pred_rf=rf_model.predict(X_test)
r2_rf=r2_score(y_test,y_pred_rf)
```
